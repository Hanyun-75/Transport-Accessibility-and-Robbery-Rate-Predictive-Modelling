## Introduction

[[ go back to the top ]](#Table-of-contents)

Crime is unevenly distributed across urban space, and public transport nodes have long been recognised as environments in which crime opportunities may concentrate. A systematic review of transit crime research suggests that stations and their surroundings can function as crime attractors, generators, absorbers, or radiators, depending on node characteristics, ridership, land use, and neighbourhood context (Ceccato, Gaudelet and Graf, 2022). This means that transport-related risk is not confined to vehicles or platforms, but often extends into station-adjacent urban space.

However, transport-node effects do not apply uniformly across all offences. Previous research suggests that transport nodes can shape crime opportunities, but these relationships vary by offence type rather than applying equally to overall crime (Irvin-Erickson and La Vigne, 2015; Li and Kim, 2023). For this reason, the study first screens several transport-relevant candidate crime types before retaining a final outcome for model comparison. Robbery is ultimately retained because it shows a clearer transport-related pattern than broader offence categories, which is consistent with previous studies linking transit nodes and accessibility to robbery or other street-crime outcomes (Li and Kim, 2023; Liu et al., 2020; Kirchmaier, Langella and Manning, 2024).

At the same time, node effects do not operate in isolation. Studies of station crime repeatedly show that surrounding neighbourhood conditions matter, including local crime levels, socioeconomic disadvantage, density, and demographic structure (Irvin-Erickson and La Vigne, 2015; Ceccato, Gaudelet and Graf, 2022). Research on London neighbourhoods similarly finds that area-level crime differences are strongly related to underlying community structure rather than any single environmental feature alone (Sutherland, Brunton-Smith and Jackson, 2013). This makes it methodologically necessary to establish a baseline using basic neighbourhood characteristics before testing transport variables.

This study therefore examines whether transport accessibility and proximity to TfL rail-based stations improve the prediction of robbery rates across London LSOAs beyond baseline neighbourhood characteristics. PTAL is used to represent broader public transport accessibility, while nearest-station distance captures simple node proximity. The aim is not to make a causal claim about transport, but to test whether transport-related variables provide additional predictive value once baseline area differences have already been taken into account.

## Research Question

## Data

[[ go back to the top ]](#Table-of-contents)

This study combines several London-wide datasets at LSOA level. Crime data comes from the MPS geographic crime breakdown. Candidate crime types are compared through descriptive transport-related gradients, after which robbery is retained as the final outcome. Robbery counts are standardised by the 2023 residential population to produce robbery rates, and the final modelling target is `log_robbery_rate`.

Baseline variables include IMD score, log population density, and the share of residents aged 16–24. Two transport variables are used: `mean_ptal_ai`, representing broader public transport accessibility, and `dist_tfl_rail_station_m`, representing proximity to the nearest TfL rail-based station. PTAL and station-location data are obtained from the Transport for London GIS Open Data Hub.

Key preprocessing steps include matching datasets through common LSOA codes, defining the final MPS-covered London sample, projecting spatial data to EPSG:27700 for area and distance calculation, and applying log transformation to strongly right-skewed variables.

## Methodology

[[ go back to the top ]](#Table-of-contents)

The analysis followed a staged supervised-learning workflow. First, the datasets were joined at LSOA level to create a common analysis sample containing 2023 crime counts, population, deprivation, PTAL accessibility, and nearest-station distance variables. Before modelling, candidate crime types were screened because transport-related crime patterns may differ by offence type. The modelling stage then compared baseline-only models with transport-extended models to test whether PTAL accessibility and station proximity added predictive information beyond neighbourhood characteristics. Model evaluation used a 75/25 train-test split, cross-validation within the training set, and held-out test performance. Finally, the selected extended tree-based model was interpreted and checked using permutation importance, partial dependence plots, residual mapping, and borough-grouped cross-validation. #### 6.3.3 Westminster sensitivity function

```python
# Check: After removing Westminster, does the station-distance gradient for ‘robbery’ still exist?
def make_distance_gradient(df, crimes):
    rows = []
    
    for crime_type in crimes:
        grouped = (
            df
            .groupby("station_distance_band", observed=True)
            .agg(
                crime_count=(crime_type, "sum"),
                population=("population_2023", "sum")
            )
        )
        
        grouped["band_rate_per_1000"] = (
            grouped["crime_count"] / grouped["population"] * 1000
        )
        
        for band, row in grouped.iterrows():
            rows.append({
                "crime_type": crime_type,
                "station_distance_band": band,
                "band_rate_per_1000": row["band_rate_per_1000"]
            })
    
    out = pd.DataFrame(rows)
    wide = out.pivot(
        index="crime_type",
        columns="station_distance_band",
        values="band_rate_per_1000"
    )
    
    return wide[["0-250m", "250-500m", "500-1000m", "1000m+"]]
```

### 6.6 Model comparison: baseline vs transport-extended

#### 6.6.1 Train-test split

```python
#1. 75/25 outer train-test split
train_idx, test_idx = train_test_split(
    model_df.index,
    test_size=0.25,
    random_state=42
)

X_base_train = model_df.loc[train_idx, baseline_features]
X_base_test  = model_df.loc[test_idx, baseline_features]

X_ext_train = model_df.loc[train_idx, extended_features]
X_ext_test  = model_df.loc[test_idx, extended_features]

y_train = model_df.loc[train_idx, target]
y_test  = model_df.loc[test_idx, target]
print("Baseline train/test:", X_base_train.shape, X_base_test.shape)
print("Extended train/test:", X_ext_train.shape, X_ext_test.shape)
print("Target train/test:", y_train.shape, y_test.shape)
```

#### 6.6.2 5-fold CV for Linear Regression and Random Forest

```python
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

models_for_cv = {
    ("Linear Regression", "Baseline"): (
        LinearRegression(),
        X_base_train
    ),
    ("Linear Regression", "Extended"): (
        LinearRegression(),
        X_ext_train
    ),
    ("Random Forest", "Baseline"): (
        RandomForestRegressor(
            n_estimators=500,
            min_samples_leaf=5,
            random_state=42,
            n_jobs=-1
        ),
        X_base_train
    ),
    ("Random Forest", "Extended"): (
        RandomForestRegressor(
            n_estimators=500,
            min_samples_leaf=5,
            random_state=42,
            n_jobs=-1
        ),
        X_ext_train
    )
}

cv_rows = []

for (model_name, feature_set), (model, X_train_use) in models_for_cv.items():
    
    scores = cross_validate(
        model,
        X_train_use,
        y_train,
        cv=cv,
        scoring={
            "r2": "r2",
            "rmse": "neg_root_mean_squared_error",
            "mae": "neg_mean_absolute_error"
        },
        n_jobs=-1
    )
    
    # sklearn returns negative error scores because higher scores are treated as better.
    # Convert them back to positive RMSE / MAE values for interpretation.
    r2_scores = scores["test_r2"]
    rmse_scores = -scores["test_rmse"]
    mae_scores = -scores["test_mae"]
    
    cv_rows.append({
        "model": model_name,
        "feature_set": feature_set,
        "cv_r2_mean": r2_scores.mean(),
        "cv_r2_std": r2_scores.std(),
        "cv_rmse_mean": rmse_scores.mean(),
        "cv_rmse_std": rmse_scores.std(),
        "cv_mae_mean": mae_scores.mean(),
        "cv_mae_std": mae_scores.std()
    })

cv_results_df = pd.DataFrame(cv_rows)

display(cv_results_df)
```

#### 6.6.3 Held-out test evaluation

```python
#Final held-out test evaluation
def evaluate_regression(model_name, feature_set, y_true, y_pred):
    return {
        "model": model_name,
        "feature_set": feature_set,
        "test_r2": r2_score(y_true, y_pred),
        "test_rmse": np.sqrt(mean_squared_error(y_true, y_pred)),
        "test_mae": mean_absolute_error(y_true, y_pred)
    }

test_results = []

# 1. Linear Regression - Baseline
lin_base = LinearRegression()
lin_base.fit(X_base_train, y_train)

y_pred_lin_base = lin_base.predict(X_base_test)

test_results.append(
    evaluate_regression(
        "Linear Regression",
        "Baseline",
        y_test,
        y_pred_lin_base
    )
)

# 2. Linear Regression - Extended
lin_ext = LinearRegression()
lin_ext.fit(X_ext_train, y_train)

y_pred_lin_ext = lin_ext.predict(X_ext_test)

test_results.append(
    evaluate_regression(
        "Linear Regression",
        "Extended",
        y_test,
        y_pred_lin_ext
    )
)

# 3. Random Forest - Baseline
rf_base = RandomForestRegressor(
    n_estimators=500,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

rf_base.fit(X_base_train, y_train)

y_pred_rf_base = rf_base.predict(X_base_test)

test_results.append(
    evaluate_regression(
        "Random Forest",
        "Baseline",
        y_test,
        y_pred_rf_base
    )
)

# 4. Random Forest - Extended
rf_ext = RandomForestRegressor(
    n_estimators=500,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

rf_ext.fit(X_ext_train, y_train)

y_pred_rf_ext = rf_ext.predict(X_ext_test)

test_results.append(
    evaluate_regression(
        "Random Forest",
        "Extended",
        y_test,
        y_pred_rf_ext
    )
)

test_results_df = pd.DataFrame(test_results)

display(test_results_df)
final_models = {
    ("Linear Regression", "Baseline"): lin_base,
    ("Linear Regression", "Extended"): lin_ext,
    ("Random Forest", "Baseline"): rf_base,
    ("Random Forest", "Extended"): rf_ext
}

final_predictions = {
    ("Linear Regression", "Baseline"): y_pred_lin_base,
    ("Linear Regression", "Extended"): y_pred_lin_ext,
    ("Random Forest", "Baseline"): y_pred_rf_base,
    ("Random Forest", "Extended"): y_pred_rf_ext
}
```

#### 6.6.4 Stage 1 summary

```python
# Combine Stage 1 CV and held-out test results
model_summary = test_results_df.merge(
    cv_results_df,
    on=["model", "feature_set"],
    how="left"
)

model_summary = model_summary[[
    "model",
    "feature_set",
    "cv_r2_mean",
    "cv_r2_std",
    "cv_rmse_mean",
    "cv_rmse_std",
    "cv_mae_mean",
    "cv_mae_std",
    "test_r2",
    "test_rmse",
    "test_mae"
]]

display(model_summary)
```

Stage 1 shows that adding transport variables improves both Linear Regression and Random Forest performance. This motivates an additional boosted-tree check using XGBoost.

#### 6.7.1 XGBoost 5-fold CV

```python
xgb_params = {
    "n_estimators": 300,
    "max_depth": 3,
    "learning_rate": 0.05,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "objective": "reg:squarederror",
    "random_state": 42,
    "n_jobs": -1
}

xgb_models_for_cv = {
    ("XGBoost", "Baseline"): (
        XGBRegressor(**xgb_params),
        X_base_train
    ),
    ("XGBoost", "Extended"): (
        XGBRegressor(**xgb_params),
        X_ext_train
    )
}

xgb_cv_rows = []

for (model_name, feature_set), (model, X_train_use) in xgb_models_for_cv.items():
    scores = cross_validate(
        model,
        X_train_use,
        y_train,
        cv=cv,
        scoring={
            "r2": "r2",
            "rmse": "neg_root_mean_squared_error",
            "mae": "neg_mean_absolute_error"
        },
        n_jobs=-1
    )

    r2_scores = scores["test_r2"]
    rmse_scores = -scores["test_rmse"]
    mae_scores = -scores["test_mae"]

    xgb_cv_rows.append({
        "model": model_name,
        "feature_set": feature_set,
        "cv_r2_mean": r2_scores.mean(),
        "cv_r2_std": r2_scores.std(),
        "cv_rmse_mean": rmse_scores.mean(),
        "cv_rmse_std": rmse_scores.std(),
        "cv_mae_mean": mae_scores.mean(),
        "cv_mae_std": mae_scores.std()
    })

xgb_cv_results_df = pd.DataFrame(xgb_cv_rows)
display(xgb_cv_results_df)
```

```python
xgb_test_rows = []
for feature_set, X_train_use, X_test_use in [
    ("Baseline", X_base_train, X_base_test),
    ("Extended", X_ext_train, X_ext_test)
]:
    xgb = XGBRegressor(**xgb_params)
    xgb.fit(X_train_use, y_train)

    y_pred = xgb.predict(X_test_use)

    xgb_test_rows.append({
        "model": "XGBoost",
        "feature_set": feature_set,
        "test_r2": r2_score(y_test, y_pred),
        "test_rmse": np.sqrt(mean_squared_error(y_test, y_pred)),
        "test_mae": mean_absolute_error(y_test, y_pred)
    })

xgb_test_results_df = pd.DataFrame(xgb_test_rows)
display(xgb_test_results_df)
```

#### 6.7.3 Consolidated model comparison

```python
xgb_summary = xgb_test_results_df.merge(
    xgb_cv_results_df,
    on=["model", "feature_set"],
    how="left"
)

xgb_summary = xgb_summary[[
    "model",
    "feature_set",
    "cv_r2_mean",
    "cv_r2_std",
    "cv_rmse_mean",
    "cv_rmse_std",
    "cv_mae_mean",
    "cv_mae_std",
    "test_r2",
    "test_rmse",
    "test_mae"
]]
model_summary_with_xgb = pd.concat(
    [model_summary, xgb_summary],
    ignore_index=True
)

# Freeze the fixed-specification results before any tuning happens
fixed_model_summary = model_summary_with_xgb.copy()

display(fixed_model_summary)
```

### 6.9 Tuning and final model selection

#### 6.9.1 Fixed extended tree-model overfitting check

```python
# RF parameters
rf_params = {
    "n_estimators": 500,
    "min_samples_leaf": 5,
    "random_state": 42,
    "n_jobs": -1
}

# XGBoost parameters
xgb_params = {
    "n_estimators": 300,
    "max_depth": 3,
    "learning_rate": 0.05,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "objective": "reg:squarederror",
    "random_state": 42,
    "n_jobs": -1
}

models_overfit_check = {
    "Random Forest Extended": (
        RandomForestRegressor(**rf_params),
        X_ext_train
    ),
    "XGBoost Extended": (
        XGBRegressor(**xgb_params),
        X_ext_train
    )
}

overfit_rows = []

for model_name, (model, X_train_use) in models_overfit_check.items():
    scores = cross_validate(
        model,
        X_train_use,
        y_train,
        cv=cv,
        scoring={
            "r2": "r2",
            "rmse": "neg_root_mean_squared_error",
            "mae": "neg_mean_absolute_error"
        },
        return_train_score=True,
        n_jobs=-1
    )
    
    train_r2 = scores["train_r2"]
    val_r2 = scores["test_r2"]
    train_rmse = -scores["train_rmse"]
    val_rmse = -scores["test_rmse"]
    train_mae = -scores["train_mae"]
    val_mae = -scores["test_mae"]
    
    overfit_rows.append({
        "model": model_name,
        "train_r2_mean": train_r2.mean(),
        "cv_r2_mean": val_r2.mean(),
        "r2_gap_train_minus_cv": train_r2.mean() - val_r2.mean(),
        "train_rmse_mean": train_rmse.mean(),
        "cv_rmse_mean": val_rmse.mean(),
        "rmse_gap_cv_minus_train": val_rmse.mean() - train_rmse.mean(),
        "train_mae_mean": train_mae.mean(),
        "cv_mae_mean": val_mae.mean(),
        "mae_gap_cv_minus_train": val_mae.mean() - train_mae.mean(),
        "cv_r2_std": val_r2.std()
    })

overfit_check_df = pd.DataFrame(overfit_rows)

display(overfit_check_df)
```

#### 6.10.4 Spatial residual diagnostic of the final model

###### 6.10.4.1 Prepare held-out test residuals

```python
# Prepare held-out test residuals for spatial diagnostic
# Residuals are calculated on the log-transformed robbery-rate scale,
# because the final model predicts log_robbery_rate.

test_pred_df = model_df.loc[
    X_ext_test.index,
    ["LSOA21CD", "LSOA21NM"]
].copy()

# Observed and predicted values on the model scale
test_pred_df["y_true_log"] = y_test.loc[X_ext_test.index]
test_pred_df["y_pred_log"] = final_model.predict(X_ext_test)

# Log-scale residual: positive means observed > predicted
test_pred_df["residual_log"] = (
    test_pred_df["y_true_log"] -
    test_pred_df["y_pred_log"]
)

# Back-transform to the original robbery-rate scale for easier interpretation
test_pred_df["observed_robbery_rate"] = np.expm1(test_pred_df["y_true_log"])
test_pred_df["predicted_robbery_rate"] = np.expm1(test_pred_df["y_pred_log"])

# Easier-to-interpret residual measures on the original / ratio scale
test_pred_df["rate_error"] = (
    test_pred_df["observed_robbery_rate"] -
    test_pred_df["predicted_robbery_rate"]
)

# Because residual_log = log(1 + observed rate) - log(1 + predicted rate),
# exp(residual_log) = (1 + observed rate) / (1 + predicted rate)
test_pred_df["rate_ratio_plus1"] = np.exp(test_pred_df["residual_log"])

# Display a compact preview
display(test_pred_df[[
    "LSOA21CD",
    "LSOA21NM",
    "observed_robbery_rate",
    "predicted_robbery_rate",
    "rate_error",
    "residual_log",
    "rate_ratio_plus1"
]].head())

# Display summary statistics for interpretation
display(test_pred_df[[
    "observed_robbery_rate",
    "predicted_robbery_rate",
    "rate_error",
    "residual_log",
    "rate_ratio_plus1"
]].describe())
```

### 6.11 Borough-grouped cross-validation sensitivity check

#### 6.11.1 Create borough groups

```python
# Spatial sensitivity check: create borough groups from LSOA names
# This is used only for grouped cross-validation within the training set.
groupcheck_df = model_df.copy()

# LSOA21NM usually has the form "Westminster 013B".
# This removes the final LSOA code and keeps the borough name.
groupcheck_df["borough_group"] = (
    groupcheck_df["LSOA21NM"]
    .str.replace(r"\s+\d{3}[A-Z]$", "", regex=True)
)

display(groupcheck_df[["LSOA21CD", "LSOA21NM", "borough_group"]].head())
print("Number of borough groups:", groupcheck_df["borough_group"].nunique())
display(groupcheck_df["borough_group"].value_counts().head(10))
```

#### 6.11.2 Borough-grouped CV for XGBoost baseline vs transport-extended models

```python
# Borough-grouped CV for XGBoost baseline vs transport-extended models
# This is a spatial sensitivity check, not the main model-selection step.

group_cv = GroupKFold(n_splits=5)

# Use fixed XGBoost specification for a fair baseline vs extended comparison.
# If xgb_fixed_params already exists earlier in the notebook, this reuses it.
try:
    xgb_params_for_groupcv = xgb_fixed_params.copy()
except NameError:
    xgb_params_for_groupcv = {
        "n_estimators": 300,
        "max_depth": 3,
        "learning_rate": 0.05,
        "subsample": 0.8,
        "colsample_bytree": 0.8,
        "objective": "reg:squarederror",
        "random_state": 42,
        "n_jobs": -1
    }

# Groups must correspond to the training rows only
groups_train = groupcheck_df.loc[X_ext_train.index, "borough_group"]

group_cv_rows = []

group_models = {
    "Baseline": X_base_train,
    "Transport-extended": X_ext_train
}

for feature_set, X_train_use in group_models.items():
    model = XGBRegressor(**xgb_params_for_groupcv)
    
    scores = cross_validate(
        model,
        X_train_use,
        y_train,
        cv=group_cv,
        groups=groups_train,
        scoring={
            "r2": "r2",
            "rmse": "neg_root_mean_squared_error",
            "mae": "neg_mean_absolute_error"
        },
        n_jobs=-1
    )
    rmse_scores = -scores["test_rmse"]
    mae_scores = -scores["test_mae"]
    
    group_cv_rows.append({
        "model": "XGBoost",
        "feature_set": feature_set,
        "group_cv_r2_mean": scores["test_r2"].mean(),
        "group_cv_r2_std": scores["test_r2"].std(),
        "group_cv_rmse_mean": -scores["test_rmse"].mean(),
        "group_cv_rmse_std": scores["test_rmse"].std(),
        "group_cv_mae_mean": -scores["test_mae"].mean(),
        "group_cv_mae_std": scores["test_mae"].std()
    })

group_cv_results_df = pd.DataFrame(group_cv_rows)

display(group_cv_results_df)
```

#### 6.11.3 Baseline-to-extended improvement under borough-grouped CV

```python
# Baseline -> Transport-extended improvement under borough-grouped CV

group_base = group_cv_results_df[
    group_cv_results_df["feature_set"] == "Baseline"
].iloc[0]

group_ext = group_cv_results_df[
    group_cv_results_df["feature_set"] == "Transport-extended"
].iloc[0]

group_cv_improvement_df = pd.DataFrame([{
    "model": "XGBoost",
    "group_cv_r2_gain": group_ext["group_cv_r2_mean"] - group_base["group_cv_r2_mean"],
    "group_cv_rmse_reduction": group_base["group_cv_rmse_mean"] - group_ext["group_cv_rmse_mean"],
    "group_cv_mae_reduction": group_base["group_cv_mae_mean"] - group_ext["group_cv_mae_mean"]
}])

display(group_cv_improvement_df)
```

Under borough-grouped cross-validation, the transport-extended XGBoost model still outperformed the baseline model. Mean grouped-CV R² increased from 0.135 to 0.421, while RMSE and MAE both decreased, indicating that the transport-variable uplift remains under a more spatially conservative validation strategy. This should still be treated as a sensitivity check rather than a full correction for spatial dependence.

## Conclusion

[[ go back to the top ]](#Table-of-contents)

This study examined whether transport accessibility and transport-node proximity improve the prediction of robbery rates across London LSOAs beyond baseline neighbourhood characteristics. The results suggest that they do. Across Linear Regression, Random Forest, and XGBoost, transport-extended models consistently outperformed baseline-only models on the held-out test set. This indicates that PTAL accessibility and nearest-station distance provide additional predictive information beyond deprivation, density, and age structure.

Robbery was retained as the final outcome because it showed a clearer and more stable transport-related pattern than the other candidate crime types. In particular, robbery displayed both a positive PTAL gradient and a clear distance-decay pattern from TfL rail-based stations, while remaining less dominated by central-area outliers than theft. The final tuned XGBoost extended model was therefore used for interpretation, but the main conclusion does not depend on a single algorithm, since the transport-extended specification improved performance across all tested model families.

At the same time, these findings should be interpreted cautiously. The analysis is conducted at LSOA level, which may smooth over finer station-area effects, and residential population is an imperfect denominator for crimes influenced by visitors, commuters, and commercial activity. In addition, PTAL and nearest-station distance are proxies for transport exposure rather than direct measures of actual movement. The study therefore supports a predictive, not causal, conclusion: transport-related variables help explain where robbery risk is higher, but they do not by themselves establish why it is higher.
